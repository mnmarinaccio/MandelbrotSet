### This Jupyter Notebook visualizes the Mandelbrot set using Python. It creates a grid of complex numbers, iterates the Mandelbrot equation for each point, and records how quickly each point "escapes" to infinity. The result is displayed as an image, where colors represent the number of iterations before escape.<br />
---

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# mandelbrot equation Zn+1 = Zn^2 + c
width, height = 800, 800
real_range = np.linspace(-2, 1, width)
imag_range = np.linspace(-1.5, 1.5, height)

#create 2d grid of complex numbers
real, imag = np.meshgrid(real_range, imag_range)
c_vals = real + 1j * imag
```

```python
maxIter = 50
escape_times = np.zeros_like(c_vals, dtype=int)

for i in range(c_vals.shape[0]):
    for j in range(c_vals.shape[1]):
        c = c_vals[i, j]
        z = 0+0j
        for n in range(maxIter):
            z=z**2+c
            if abs(z) > 2:
                escape_times[i, j] = n
                break
  
```

```python
# Plot Mandelbrot Set
plt.figure(figsize=(10, 10))
plt.imshow(escape_times, extent=[-2, 1, -1.5, 1.5], cmap='magma', origin='lower')
plt.colorbar(label='Iterations to Escape')
plt.title('Mandelbrot Set')
plt.xlabel('Real Part')
plt.ylabel('Imaginary Part')
plt.show()
```
